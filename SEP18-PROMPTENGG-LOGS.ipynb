```python
def call_llm(prompt,model="gpt-4o"):
  from openai import OpenAI

  client = OpenAI()

  completion = client.chat.completions.create(
    model=model,
    messages=[
     {"role": "user", "content": prompt}
    ]
  )

  print(completion.choices[0].message.content)
```

```python
call_llm("Hey can you tell me the usecase of ai in two lines!")
```

```python
def get_completion(prompt, model="gpt-4o"):
    from openai import OpenAI

    openai_client = OpenAI()

    # Construct the message list with a single user message
    messages = [
        {
            "role": "user",
            "content": prompt
        }
    ]

    # Make the API call to OpenAI's chat completion endpoint
    response = openai_client.chat.completions.create(
        model=model,          # Specify which model to use
        messages=messages,    # Message history provided to the model
        temperature=0         # 0 = deterministic, 1 = more random
    )

        # Extract and return the generated response
    return response.choices[0].message.content

```

```python
get_completion("Hey can you tell me the usecase of ai in two lines!")
```
